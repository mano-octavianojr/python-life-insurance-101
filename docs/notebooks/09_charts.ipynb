## Insurance in everyday words

Total premiums paid is the sum of payments. It is not the death benefit or a savings balance. These examples are term policies; a benefit depends on a covered death and policy terms. The graph cannot decide whether a policy is suitable.

## Python toolbox

`range(0, 11)` generates years 0 through 10. A list comprehension builds a list with one calculation per year. matplotlib draws a figure: label both axes, include units, and add a legend. Assume premiums stay fixed and all scheduled payments are made.

## Predict, then run

At year 10, where will the lines for ₱250/month and ₱350/month end?

```python
import matplotlib.pyplot as plt
years = list(range(0, 11))
oak_paid = [250 * 12 * year for year in years]
river_paid = [350 * 12 * year for year in years]
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(years, oak_paid, marker="o", label="Oak: PHP 250/month")
ax.plot(years, river_paid, marker="s", label="River: PHP 350/month")
ax.set(title="Fictional term policies: cumulative premiums", xlabel="Years of payments", ylabel="Total premiums paid (PHP)")
ax.legend()
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

### Read selected points before interpreting the chart

Predict the totals and their difference at years 0, 2, and 10.

```python
for selected_year in [0, 2, 10]:
    selected_difference = river_paid[selected_year] - oak_paid[selected_year]
    print(f"Year {selected_year}: Oak PHP {oak_paid[selected_year]:,}, "
          f"River PHP {river_paid[selected_year]:,}, difference PHP {selected_difference:,}")
assert river_paid[2] - oak_paid[2] == 2400
```

**What happened and why:** The differences are PHP 0, PHP 2,400, and PHP 12,000. They grow by PHP 1,200 per year because the fixed monthly payments differ by PHP 100. The index equals the year only because this list begins at year zero and advances one year at a time.

### Plot annual payments instead of cumulative payments

Predict whether the annual-payment lines slope upward. These are the same fixed prices but a different quantity.

```python
annual_years = list(range(1, 11))
annual_oak = [250 * 12 for year in annual_years]
annual_river = [350 * 12 for year in annual_years]
annual_fig, annual_ax = plt.subplots(figsize=(8, 4))
annual_ax.plot(annual_years, annual_oak, marker="o", label="Oak")
annual_ax.plot(annual_years, annual_river, marker="s", label="River")
annual_ax.set(title="Fictional term policies: payments each year",
              xlabel="Payment year", ylabel="Annual premiums paid (PHP)", ylim=(0, 5000))
annual_ax.legend()
annual_ax.grid(alpha=0.3)
annual_fig.tight_layout()
plt.show()
assert annual_oak == [3000] * 10 and annual_river == [4200] * 10
```

**What happened and why:** The lines are horizontal at PHP 3,000 and PHP 4,200. Annual payments stay fixed while cumulative payments keep adding up. Starting the vertical scale at zero gives context to the payment amounts; always read the axis label before interpreting a line.
