## Insurance in everyday words

Total premiums paid is the sum of payments. It is not the death benefit or a savings balance. These examples are term policies; a benefit depends on a covered death and policy terms. The graph cannot decide whether a policy is suitable.

## Python toolbox

`range(0, 11)` generates years 0 through 10. A list comprehension builds a list with one calculation per year. matplotlib draws a figure: label both axes, include units, and add a legend. Assume premiums stay fixed and all scheduled payments are made.

## Predict, then run

At year 10, where will the lines for ₱250/month and ₱350/month end?

```python
import matplotlib.pyplot as plt
years = list(range(0, 11))
oak_paid = [250 * 12 * year for year in years]
river_paid = [350 * 12 * year for year in years]
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(years, oak_paid, marker="o", label="Oak: PHP 250/month")
ax.plot(years, river_paid, marker="s", label="River: PHP 350/month")
ax.set(title="Fictional term policies: cumulative premiums", xlabel="Years of payments", ylabel="Total premiums paid (PHP)")
ax.legend()
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
