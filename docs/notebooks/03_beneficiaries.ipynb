## Insurance in everyday words

A beneficiary is a person chosen to receive the insurance money. Naming beneficiaries, changes, and payment rules depend on the policy and local law. This classroom example splits an approved benefit using specified percentages.

## Python toolbox

A list stores items in order; index 0 selects the first item. A dictionary pairs keys with values. `person["share"]` reads a value by its key. Divide a percentage by 100 before multiplying.

## Predict, then run

For a ₱500,000 benefit, what will shares of 60% and 40% produce?

```python
benefit = 500000
beneficiaries = [{"name": "Ana", "share": 60}, {"name": "Ben", "share": 40}]
first = beneficiaries[0]
second = beneficiaries[1]
first_amount = benefit * first["share"] / 100
second_amount = benefit * second["share"] / 100
print(first["name"], f"₱{first_amount:,.0f}")
print(second["name"], f"₱{second_amount:,.0f}")
```
