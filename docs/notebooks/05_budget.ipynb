## Story opening

Maya and Leo create a fictional household budget. They want to see how an insurance premium fits alongside ordinary expenses without using anyone’s private information.

## Learning goals

Loop through a dictionary; accumulate totals; calculate a remainder.

## Insurance in everyday words

Premiums are one household expense. A budget compares money coming in with money going out. Money left after these listed expenses is not necessarily freely available: real households may have other costs.

## Python toolbox

`for` repeats instructions. `.items()` gives each dictionary key and value. `+=` adds to the current total. The loop body must be indented.

## Predict, then run

With income of ₱20,000, will the listed expenses leave a positive remainder?

```python
income = 20000
expenses = {"food": 6500, "transport": 2000, "housing": 8000, "premium": 350}
total = 0
for category, amount in expenses.items():
    print(category, amount)
    total += amount
remaining = income - total
print(f"Listed expenses: ₱{total:,}; remaining: ₱{remaining:,}")
```

Use .copy() to preserve the original. Assign a new value with a dictionary key. sum() adds a collection of numbers.

## Sample solution

### Trace the running total

Leo uses a smaller list of expenses to see what += does. Predict the total after each loop step.

```python
practice_expenses = {"food": 6000, "transport": 1800, "premium": 400}
practice_total = 0
for category, amount in practice_expenses.items():
    practice_total += amount
    print(f"After {category}: PHP {practice_total:,}")
assert practice_total == 8200
```

**What happened and why:** The running totals are PHP 6,000, PHP 7,800, and PHP 8,200. The accumulator starts at zero before the loop; initializing it inside the loop would discard earlier amounts.

### Model an extra expense without changing the original

Use the original fictional budget. Predict the remainder after adding a PHP 4,000 repair expense.

```python
repair_budget = expenses.copy()
repair_budget["repair"] = 4000
repair_total = sum(repair_budget.values())
repair_remainder = income - repair_total
print(f"With repair: PHP {repair_total:,} in listed expenses")
print(f"Remainder: PHP {repair_remainder:,}")
assert repair_remainder == -850
assert "repair" not in expenses
```

**What happened and why:** The listed expenses become PHP 20,850 and the remainder is negative PHP 850. A negative result describes a shortfall in this model; the program cannot decide which expenses a household should change.
