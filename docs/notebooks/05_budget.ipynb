## Story opening

Maya and Leo create a fictional household budget. They want to see how an insurance premium fits alongside ordinary expenses without using anyone’s private information.

## Learning goals

Loop through a dictionary; accumulate totals; calculate a remainder.

## Insurance in everyday words

Premiums are one household expense. A budget compares money coming in with money going out. Money left after these listed expenses is not necessarily freely available: real households may have other costs.

## Python toolbox

`for` repeats instructions. `.items()` gives each dictionary key and value. `+=` adds to the current total. The loop body must be indented.

## Predict, then run

With income of ₱20,000, will the listed expenses leave a positive remainder?

```python
income = 20000
expenses = {"food": 6500, "transport": 2000, "housing": 8000, "premium": 350}
total = 0
for category, amount in expenses.items():
    print(category, amount)
    total += amount
remaining = income - total
print(f"Listed expenses: ₱{total:,}; remaining: ₱{remaining:,}")
```

Use .copy() to preserve the original. Assign a new value with a dictionary key. sum() adds a collection of numbers.

## Sample solution

```python
revised = expenses.copy()
revised["transport"] = 2500
revised["school supplies"] = 700
revised_total = sum(revised.values())
revised_remaining = income - revised_total
print(f"New remainder: ₱{revised_remaining:,}")
```
