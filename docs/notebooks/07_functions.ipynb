## Insurance in everyday words

A simplified coverage gap is an estimated money need minus resources available to meet that same need, with a minimum of zero. For this fictional family, the need is the money their dependents would require after a death, and resources are savings set aside for that purpose. Do not count money that is unavailable or counted twice. This exercise ignores inflation, debts changing over time, taxes, and many other factors. It is not a recommendation or a way to calculate an insurance price.

## Python toolbox

Use finite, nonnegative numeric inputs in this introductory calculator; visitor-text validation comes in lesson 10. `def` starts a function. Parameters name its inputs. `return` sends a result back. `max(0, value)` keeps a negative result from becoming a negative gap. `raise ValueError` reports an unacceptable input; try/except can handle that expected error.

## Predict, then run

What gap results from a ₱600,000 need and ₱200,000 resources? What if resources exceed the need?

```python
def annual_premium(monthly):
    if monthly < 0:
        raise ValueError("Monthly premium cannot be negative")
    return monthly * 12

def coverage_gap(need, resources):
    if need < 0 or resources < 0:
        raise ValueError("Use nonnegative amounts")
    return max(0, need - resources)

print(annual_premium(350))
print(coverage_gap(600000, 200000))
print(coverage_gap(100000, 150000))
```

### Use returned values in another calculation

Predict the annual totals for PHP 250 and PHP 350 per month, then their difference.

```python
oak_annual_example = annual_premium(250)
river_annual_example = annual_premium(350)
annual_difference_example = river_annual_example - oak_annual_example
print("Annual totals:", oak_annual_example, river_annual_example)
print("Annual difference:", annual_difference_example)
assert annual_difference_example == 1200
```

**What happened and why:** The functions return 3,000 and 4,200, so the difference is 1,200. return makes a value available to the caller; printing a value alone would not let this subtraction use it.

### Compare resources while holding need fixed

Assume all resources are available for the same fictional need and counted once. Predict how the gap changes as resources rise.

```python
scenario_need = 600000
for scenario_resources in [100000, 400000, 700000]:
    scenario_gap = coverage_gap(scenario_need, scenario_resources)
    print(f"Resources PHP {scenario_resources:,}: gap PHP {scenario_gap:,}")
assert coverage_gap(600000, 100000) == 500000
assert coverage_gap(600000, 400000) == 200000
assert coverage_gap(600000, 700000) == 0
```

**What happened and why:** The gaps are PHP 500,000, PHP 200,000, and zero. max() stops the last result becoming negative. Zero says the supplied resources meet the supplied need in this model, not that every financial risk is resolved.
