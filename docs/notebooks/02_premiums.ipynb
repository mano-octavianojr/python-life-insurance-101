## Python toolbox

A variable is a name for a value. `=` assigns a value; `*` multiplies and `-` subtracts. An f-string starts with `f` and inserts values inside braces. `:,` adds thousands separators.

## Predict, then run

Will 12 monthly payments of ₱350 cost more or less than a fictional annual payment of ₱4,000?

```python
monthly_premium = 350
annual_option = 4000
monthly_total = monthly_premium * 12
saving = monthly_total - annual_option
print(f"Monthly schedule total: ₱{monthly_total:,}")
print(f"Annual schedule total: ₱{annual_option:,}")
print(f"Difference: ₱{saving:,}")
```

### Compare a six-month payment schedule

Assume equal coverage for one year and all payments made. Predict whether two payments of PHP 2,100 cost more than twelve payments of PHP 375.

```python
six_month_payment = 2100
six_month_total = six_month_payment * 2
comparison_monthly_total = 375 * 12
schedule_difference = comparison_monthly_total - six_month_total
print(f"Six-month schedule: PHP {six_month_total:,}")
print(f"Monthly schedule: PHP {comparison_monthly_total:,}")
print(f"Difference: PHP {schedule_difference:,}")
assert schedule_difference == 300
```

**What happened and why:** The totals are PHP 4,200 and PHP 4,500, a PHP 300 difference. Multiply by the number of payments, not by the number of months between payments.

### Estimate a monthly set-aside for an annual payment

A fictional annual payment is PHP 4,800. Predict how much would need to be set aside each month over twelve months.

```python
annual_payment_example = 4800
monthly_set_aside = annual_payment_example / 12
print(f"Monthly set-aside: PHP {monthly_set_aside:,.2f}")
assert monthly_set_aside == 400
```

**What happened and why:** Dividing gives PHP 400 per month. :,.2f formats two decimal places. Setting money aside is a budget calculation, not an insurer-approved payment schedule; it does not make the annual payment due later.
