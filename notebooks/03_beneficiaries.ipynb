## Insurance in everyday words

A beneficiary is a person or organization chosen to receive the insurance money. Our examples use people. Naming beneficiaries, changes, and payment rules depend on the policy and local law. This classroom example splits an approved benefit using specified percentages.

## Python toolbox

A list stores items in order; index 0 selects the first item. A dictionary pairs keys with values. `person["share"]` reads a value by its key. `/` divides; `:,.0f` displays a number with thousands separators and no decimal places. These examples use whole-peso results; real payment rounding follows the policy and payment rules. Divide a percentage by 100 before multiplying.

## Predict, then run

For a ₱500,000 benefit, what will shares of 60% and 40% produce?

```python
benefit = 500000
beneficiaries = [{"name": "Ana", "share": 60}, {"name": "Ben", "share": 40}]
first = beneficiaries[0]
second = beneficiaries[1]
first_amount = benefit * first["share"] / 100
second_amount = benefit * second["share"] / 100
print(first["name"], f"₱{first_amount:,.0f}")
print(second["name"], f"₱{second_amount:,.0f}")
```

### Split a benefit equally three ways

Predict each amount for three fictional beneficiaries sharing a PHP 900,000 approved benefit. Here we calculate an equal split directly instead of rounding 100 / 3 to a percentage.

```python
equal_names = ["Ana", "Ben", "Cora"]
equal_benefit = 900000
equal_amount = equal_benefit / 3
print(equal_names[0], f"PHP {equal_amount:,.0f}")
print(equal_names[1], f"PHP {equal_amount:,.0f}")
print(equal_names[2], f"PHP {equal_amount:,.0f}")
assert equal_amount * 3 == equal_benefit
```

**What happened and why:** Each receives PHP 300,000 in this classroom calculation. Indexes 0, 1, and 2 retrieve the three names. Calculating an equal split directly avoids introducing rounded percentages that do not total exactly 100.

### Spot an incomplete allocation

Maya reads proposed shares of 45% and 45%. Predict the total shares and the amount still unallocated.

```python
proposed = [{"name": "Dan", "share": 45}, {"name": "Eva", "share": 45}]
proposed_share_total = proposed[0]["share"] + proposed[1]["share"]
unallocated_share = 100 - proposed_share_total
unallocated_amount = 600000 * unallocated_share / 100
print(f"Proposed shares: {proposed_share_total}%")
print(f"Unallocated in this example: PHP {unallocated_amount:,.0f}")
assert proposed_share_total == 90 and unallocated_amount == 60000
```

**What happened and why:** The shares total 90%, leaving 10%, or PHP 60,000, unallocated in the arithmetic model. The program reveals an issue; it does not choose a recipient or determine what an insurer would do.
