## Insurance in everyday words

A simplified coverage gap is an estimated money need minus resources available to meet that same need, with a minimum of zero. For this fictional family, the need is the money their dependents would require after a death, and resources are savings set aside for that purpose. Do not count money that is unavailable or counted twice. This exercise ignores inflation, debts changing over time, taxes, and many other factors. It is not a recommendation or a way to calculate an insurance price.

## Python toolbox

Use finite, nonnegative numeric inputs in this introductory calculator; visitor-text validation comes in lesson 10. `def` starts a function. Parameters name its inputs. `return` sends a result back. `max(0, value)` keeps a negative result from becoming a negative gap. `raise ValueError` reports an unacceptable input; try/except can handle that expected error.

## Predict, then run

What gap results from a ₱600,000 need and ₱200,000 resources? What if resources exceed the need?

```python
def annual_premium(monthly):
    if monthly < 0:
        raise ValueError("Monthly premium cannot be negative")
    return monthly * 12

def coverage_gap(need, resources):
    if need < 0 or resources < 0:
        raise ValueError("Use nonnegative amounts")
    return max(0, need - resources)

print(annual_premium(350))
print(coverage_gap(600000, 200000))
print(coverage_gap(100000, 150000))
```
