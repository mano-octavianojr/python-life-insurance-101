## Insurance in everyday words

## Python toolbox

Functions can call other functions. `float()` converts numeric text; reject nonfinite values such as nan and infinity using math.isfinite. Raise a clear error for invalid input. In this notebook, editable variables stand in for visitor input so Run All never waits for typing. `.lower()` and `.strip()` normalize quiz answers. `isinstance` checks a value’s type; `zip` pairs each question with an answer; `int(True)` is 1 and `int(False)` is 0, which lets us count correct answers.

## Predict, then run

For budget text "400", which policies appear? What should happen for "-1" or "hello"?

```python
import math
policies = [
    {"name": "Oak", "monthly": 250, "coverage": 300000, "term": 10},
    {"name": "River", "monthly": 350, "coverage": 500000, "term": 10},
    {"name": "Sky", "monthly": 600, "coverage": 800000, "term": 20},
]

def parse_budget(text):
    try:
        amount = float(text)
    except (ValueError, TypeError):
        raise ValueError("Enter a number, such as 400") from None
    if not math.isfinite(amount) or amount < 0:
        raise ValueError("Budget must be a finite, nonnegative number")
    return amount

def find_policies(budget_text, minimum_coverage=0):
    budget = parse_budget(budget_text)
    if not isinstance(minimum_coverage, (int, float)) or not math.isfinite(minimum_coverage) or minimum_coverage < 0:
        raise ValueError("Minimum coverage must be finite and nonnegative")
    matches = []
    for policy in policies:
        if policy["monthly"] <= budget and policy["coverage"] >= minimum_coverage:
            matches.append(policy)
    return matches

visitor_budget = "400"  # Edit this fictional input; no personal data needed.
for policy in find_policies(visitor_budget):
    annual = policy["monthly"] * 12
    full_term = annual * policy["term"]
    print(f"{policy['name']}: annual premiums ₱{annual:,}, full-term premiums ₱{full_term:,}, covered-death benefit ₱{policy['coverage']:,}")
print("Totals assume fixed premiums and every scheduled payment; not a policy recommendation.")

quiz = [
    {"question": "What is an insurance payment called?", "answer": "premium"},
    {"question": "Who is chosen to receive the benefit?", "answer": "beneficiary"},
    {"question": "Does a budget match guarantee eligibility? yes/no", "answer": "no"},
]
visitor_answers = ["premium", "beneficiary", "no"]  # Change to try the quiz.
def grade_quiz(answers):
    if len(answers) != len(quiz):
        raise ValueError("Provide one answer for each question")
    score = 0
    for question, answer in zip(quiz, answers):
        if not isinstance(answer, str):
            raise ValueError("Quiz answers must be text")
        correct = answer.strip().lower() == question["answer"]
        print(question["question"], "Correct" if correct else "Review: " + question["answer"])
        score += int(correct)
    return score
print("Quiz score:", grade_quiz(visitor_answers), "/", len(quiz))
```
