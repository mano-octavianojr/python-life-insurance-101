## Learning goals

## Insurance in everyday words

The coverage amount is the benefit specified for a covered death. A premium is the payment, not the coverage amount. Policies can also differ in exclusions (situations not covered), payment rules, eligibility (who can qualify), and renewal terms (rules for continuing coverage); this small table omits those details.

## Python toolbox

Each list item is a policy dictionary. Read fields by key. An if statement inside a loop can select matching records. `.append()` adds an item to a list.

## Predict, then run

Which policies fit a monthly budget of ₱400?

```python
policies = [
    {"name": "Oak", "monthly": 250, "coverage": 300000, "term": 10},
    {"name": "River", "monthly": 350, "coverage": 500000, "term": 10},
    {"name": "Sky", "monthly": 600, "coverage": 800000, "term": 20},
]
budget = 400
matches = []
for policy in policies:
    print(f"{policy['name']}: ₱{policy['monthly']}/month, ₱{policy['coverage']:,} benefit, {policy['term']} years")
    if policy["monthly"] <= budget:
        matches.append(policy["name"])
print("Within this budget:", matches)
```
