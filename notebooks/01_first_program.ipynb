## Insurance in everyday words

Life insurance is an agreement with an insurer (the insurance company). Someone pays premiums (payments for coverage), and the insurer pays a benefit (the insurance money) if the insured person dies while covered, subject to the policy terms (the rules in the insurance agreement). The insured person is the person whose life is covered. The person paying and the insured person need not be the same.
