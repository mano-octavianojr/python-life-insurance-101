## Python toolbox

`date(year, month, day)` creates a calendar date. `<` means earlier than and `>=` means on or after. Comparisons give True or False. Indent the instructions inside each conditional branch by four spaces.

## Predict, then run

Is the exact end date inside our stated window?

```python
from datetime import date
start = date(2026, 1, 1)
end = date(2036, 1, 1)
event_date = date(2030, 6, 15)
if event_date < start:
    status = "Before the example window"
elif event_date >= end:
    status = "After or at the end of the example window"
else:
    status = "Inside the example window; this is not a claim decision"
print(status)
```

### Inspect the last day before the end

Under the classroom rule start <= candidate < end, predict whether December 31, 2035 is included.

```python
last_day_example = date(2035, 12, 31)
last_day_inside = start <= last_day_example < end
print("Last day before end is inside:", last_day_inside)
print("Exact end is inside:", start <= end < end)
assert last_day_inside is True and (start <= end < end) is False
```

**What happened and why:** The day before the end is included; the exact end is excluded. A chained comparison requires both conditions to be true. These are date checks under our convention, not claims decisions.

### Check a fictional renewal window separately

A second example window begins at the first window's end. Predict which window contains January 1, 2036. Assume the second window is supplied; the code does not establish renewal.

```python
renewal_start = date(2036, 1, 1)
renewal_end = date(2046, 1, 1)
renewal_candidate = date(2036, 1, 1)
in_original_window = start <= renewal_candidate < end
in_renewal_window = renewal_start <= renewal_candidate < renewal_end
print("Original window:", in_original_window)
print("Supplied renewal window:", in_renewal_window)
assert (in_original_window, in_renewal_window) == (False, True)
```

**What happened and why:** The date is outside the original window and inside the supplied second window. Including starts and excluding ends avoids counting the shared boundary in both adjacent windows.
